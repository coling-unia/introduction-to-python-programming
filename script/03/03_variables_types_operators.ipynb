# Variables, Types and Operators

In this chapter, we will learn how to use variables, types, and operators in Python.

## Imperative Programming Paradigm

**Imperative**: *First do this, then do this.*  
* Imperative porgramming means _procedural programming_.
* Control Structures execute computational steps, and the state of the program changes as a function of time.
* Commands can be grouped into procedures.

- Example:

    ``Celsius_to_Fahrenheit(c)``
    1) Multiply ``c`` with 1.8 and save the result to ``temp``.
    2) Add 32 to ``temp`` and return the result of this.

## Elements of Imperative Programs  

Imperative programming is based on a sequence of instructions that change the program’s state. The key elements of imperative programs include:  

- **Variables**: Used to store and manage data during program execution.  
- **Assignments**: Allow the modification of variable values, enabling state changes.  
- **Expressions**: Represent computations that produce values, often involving variables and operators.  
- **Control Structures**: Include loops and branches, which define the flow of execution by enabling repetition and decision-making.

In the following, we will look at each of these components in detail.

## Values, Variables, Data Types

- **Values** may have different **data types**: numbers, lists, strings, etc.

- **Variables** are **placeholders** for values.


- **Variables** point to positions in memory where values are stored. 
The value of a variable can change over time (it can point to a different location or overwrite the memory location's value).

### Some data types in Python:

### Dynamic Typing

Variables in Python do not have fixed data types.

- The type of a variable is the assigned value's data type.
- During runtime, a variable can take values of different types.

    ```py
    >>> x = 15.4
    >>> type(x)
    <class 'float'>

    >>> x = "Python is great!"
    >>> type(x)
    <class 'str'>
    ```


### Floating Point Numbers

- Decimal numbers are represented as **floats** (`1.1`, `47.11`).
- Range depends on the system.
- **CAREFUL!** Often, the internal representation of floating point numbers is imprecise.

    ``` py
    >>> 0.01
    0.10000000000000001
    ```

- What to do about this? ⇒ Use ``ε`` when comparing floating point numbers.

    ``` py
    >>> epsilon = 0.0000000000000001
    >>> x_equal_y = abs(x - y) < epsilon
    >>> x_equal_y
    True
    ```

### Precedence

- Expressions may contain multiple operators: `3 + 2*4`
- **Precedence** = order in which operators are evaluated.
- Standard precedence rules: multiplication/division before addition/subtraction.
- Parentheses indicate precedence directly.

    Example:
    - ``3 + (2 * 4) = 11``

    - ``(3 + 2) * 4 = 20``

- Don't use parentheses when precedence is irrelevant. For example, `2 + 3 + 4` is better than `2 + (3 + 4)`.
- **Style**: Sometimes, it is recommended to use parentheses even if they are not strictly necessary for better legibility.

### Assignments

1. `var = expr`

    The expression `expr` is evaluated, then its value is stored in `var`.

2. `var1 = var2 = ... = expr`

    The value of `expr` is assigned to all variables `var_i` (all variables point to the <u>same value</u> in memory).


````python
a = b = 6.0 / 4.0
print(a)  # Output: 1.5
print(b)  # Output: 1.5
````

3. `var1, var2, ..., varn = expr1, expr2, ..., exprn`

    All `expr_i` are evaluated first, then the corresponding values are assigned to `var_i`.

````python
a, b = 6.0 / 4.0, 'Hello' + 'World'
print(a)  # Output: 1.5
print(b)  # Output: HelloWorld
````

In general, use option (1). Options (2) and (3) are sometimes used in more complex cases (later in the course!).

### Assignments Shorthands

| Long Version       | Shorthand  |
|--------------------|-----------|
| `x = x + expr`    | `x += expr` |
| `x = x - expr`    | `x -= expr` |
| `x = x * expr`    | `x *= expr` |
| `x = x / expr`    | `x /= expr` |
| `x = x % expr`    | `x %= expr` |

````python
a = 5
a += 3
print(a)  # Output: 8

a /= 2
print(a)  # Output: 4
````

### Type Conversion in Python

Reading a String from the Console: `variable = input(prompt)`

- Anything we read from the console is a **string**, even if we enter `'42.0`'.
- For computation, we need to convert this to some **number**.

    ```py
    float_number = float(variable)
    ```

- Conversion of numbers (or other types) to strings works like this: `text = str(number)`
