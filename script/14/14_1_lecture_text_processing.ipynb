### The `re` Module - Important Methods

````python
import re

string = 'Woodchucks woodchucks wudchugz'
pattern = '[Ww]oodchucks?'

re.match(pattern, string) # from beginning
re.search(pattern, string) # anywhere (first)
re.findall(pattern, string) # anywhere (all)

re.split(pattern, string, maxsplit=0)

repl = 'groundhog'
re.sub(pattern, repl, string, count=0) # substitute all
````

It better to compile the regular expression, because it makes the code more readable and much more efficient.

````python
string = 'Woodchucks woodchucks wudchugz'
pattern = '[Ww]oodchucks?'
pattern = re.compile(pattern) # compile first!!

pattern.search(string)
pattern.match(string)
pattern.findall(string)

pattern.split(string, maxsplit=0)

repl = 'groundhog'
pattern.sub(repl, string, count=0)
````

### Excursuse: Raw Strings and Format Strings

````python
print('normal string\n with \\ backslash \\n')
# normal string
# with \ backslash \n

print(r'raw string\n with \ backslash \n')
# raw string\n with \ backslash \n

var = 2; print(f'format string with var {var}')
# format string with var 2
````

````python
import math
var = 2

print(f'format string with var {var} and {math.pi}')
# format string with var 2 and 3.141592653589793

print(f'format string with var {var:.2f}'
      f' and {math.pi:.2f}')
# format string with var 2.00 and 3.14

table = {'Sjoerd': 4127, 'Jack': 4098, 'Saoirse': 767}
for name, phone in table.items():
    print(f'{name:10} ==> {phone:10d}')
# Sjoerd     ==>       4127
# Jack       ==>       4098
# Saoirse    ==>        767
````

### RegEx Errors

Find me all instances of the word "the" in a text.

* `the`
  → Misses capitalized examples
* `[tT]he`
  → Incorrectly returns “other” or “theology”
* `[^a-zA-Z][tT]he[^a-zA-Z]`


The process we just went through was based on fixing two kinds of errors:
1. Matching strings that we should not have matched (there, then, other):
   False positives (Type I errors)
2. Not matching things that we should have matched (The):
   False negatives (Type II errors)

In machine learning, we are always dealing with these kinds of errors.
Reducing the error rate for an application often involves two opposing efforts:
* Increasing accuracy or precision (minimizing false positives)
* Increasing coverage or recall (minimizing false negatives)
