### Reading input from a file

Finally, we can open and read files, so that we can use their contents as input to our functions.

````python
def grep(filename, word):
    '''Returns True if filename contains word'''
    f = open(filename)
    while True:
        line = f.readline()
        if line == '': # no input -> stop
            break
        if word in line:
            return True
    f.close()
    return False
````

- `open(filename)`
    * opens a file for reading
    * returns a "file object"
- `f.readline()`
    * reads one line from file object `f`
    * the empty string (`''`) indicates the end of file
- `f.close()`
    * closes file object `f`
    * no further file operations possible
    * memory can be freed up

### Opening a file

- `open(filename, mode)`
    * opens a file for reading or writing (depending on mode)
- `mode` is a string
    * `'r'`: open the file for reading (the default)
    * `'w'`: open the file for writing, overwriting old contents
    * `'a'`: open the file for writing, appending to the end
    * `'t'`: text mode (the default)
    * `'b'`: binary mode

### More file operations

- `f.write(somestring)`
    * writes the string `somestring` to the file `f`
- `f.read([size])`
    * reads at most size characters from file `f`
    * ... or the complete file if size not specified
- `f.readlines()`
    * returns a list of strings (= lines of file `f`)

### Resource Management

- Always close a file object when you are done with it!
- Why? Resources are limited!
    * upper limit on open files
    * conflicts when multiple processes try to change the same file

What's wrong with the follow code?

````python
def grep(filename, word):
    '''Returns True if filename contains word'''
    f = open(filename)
    while True:
        line = f.readline()
        if line == '': # no input -> stop
            break
        if word in line:
            return True
    f.close()
    return False
````

### The `with` Statement

- Always close a file object when you are done with it!
- The `with` statement is a convenient way to do this automatically:

````python
def grep(filename, word):
    '''Returns True if filename contains word'''
    with open(filename) as f:
        while True:
            line = f.readline()
            if line == '': # no input -> stop
                break
            if word in line:
                return True
    return False
````

- `with open(filename) as var`
    * opens the file `filename`
    * assigns the corresponding file object to `var`
    * automatically closes the file when we leave the `with`-block
- More generally:
    * `with` can be used with any type of "context manager"
    * Context managers are useful to automatically trigger certain actions upon entering or leaving the `with`-block

### Reading input with `for`-loops

- File objects can be used in `for`-loops
    * in each iteration step, we read one line of the input file

````python
def grep(filename, word):
    '''Returns True if filename contains word'''
    with open(filename) as f:
        for line in f:
            if word in line:
                return True
    return False
````

### Exceptions

The following code throws an error:

````python
import sys

def main():
    words = 0
    with open('ThisFileDoesNotExist.txt') as f:
        for line in f:
            words += len(line.split())
        print(words, 'words')


main()
````

- Exceptions are errors that occur at runtime
    * usually result in an error message
    * Python stops executing the program :(
- Some errors are fatal
    * stopping the program ist the only thing we can do
- However, not all errors are fatal. If possible, they should be explicitly handled within the program itself.

Take the following example:

````python
def main():
    words = 0
    try:
        with open(sys.argv[1]) as f:
            for line in f: words += len(line.split())
    except FileNotFoundError:
        print('Cannot open file:', sys.argv[1])
    except IndexError:
        print('No input file specified!')
    else:
        print(words, 'words')

if __name__ == '__main__':
    main()
````

As you can see, Python throws different exceptions with decriptive names, depending on the the type of error that occured (e.g. file not found or attempted access to a list index that does not exist).
And we can use these different exception types to define specialized behavior for each potential issue.

### Catching Exceptions

We can catch and handle exceptions with `try`-`except`-blocks. The code that might throw the exception(s) needs to be in a `try`-block, so that one or more `except`-blocks can be defined for each expects exception type. 

An `except` without a specific exception type can come after all the specific ones to catch all other exception types. This can be useful for debugging, but should be avoided in practice, because you will miss when a new unexpected exception occurs, which easily leads to even more unexpected behavior.

An optional `else` block is entered if no exception occured.

````python
try:
    # statements that can cause exceptions
    ...
except Exception_1:
    # handle exceptions of type Exception_1
    ...
except Exception_k:
    # handle exceptions of type Exception_k
    ...
except:
    # handle any other exception (NOT RECOMMENDED!)
    ...
else:
    # executed if try-block caused no exceptions
    ...
finally:
    # always executed, clean-up code
    ...
````

The optional `finally` block is guaranteed to be executed, no matter if an exception was thrown or not and whether or not all thrown exceptions were caught:

````python
try:
    assert True
except:
    print('exception')
finally:
    print('nothing ever happened')
print('this tries to pretend nothing ever happened and looks like it does the same as "finally"')
````

````python
try:
    assert False
except:
    print('exception')
finally:
    print('something happened and got caught, this still gets printed')
print('this tries to pretend nothing ever happened')
````

````python
try:
    assert False
finally:
    print('something happened and did not get caught, this still gets printed')
print('this tries to pretend nothing ever happened, but fails')
````

Exception types are not mutually exclusive.
- e.g. `FileNotFoundError` is a subclass of `Exception`, `ZeroDivisionError` is a subclass of `ArithmeticError`
- In the end, every exception is a subclass of `Exception`.
- The first applicable exception handler is executed.

Another example:

````python
def incr(d, k):
    '''Adds 1 to value of key k in dict d'''
    try:
        d[k] += 1
    except KeyError:
        d[k] = 1
````

### The `raise` Statement

Use the `raise` statement to throw a specific exception yourself, to communicate to users and other programmers how to properly use your code.

````python
def find(pairs, key):
    for (k, v) in pairs:
        if k == key:
            return v
    raise KeyError

print(find([('a',1), ('b',2), ('c',3)], 'b'))
# prints 2
print(find([('a',1), ('b',2), ('c',3)], 'f'))
# throws KeyError: 'f'
````

### Encodings

- Strings are internally represented as sequences of bytes (numbers)
- File contents are stored in a specific _encoding_
    * Encoding = mapping between numbers and characters
    * "encode": go from characters (str) to bytes/numbers
    * "decode": go from bytes to characters
- Default value of `encoding` keyword arg of `open()` is system-dependent
    * --> better to specify

````python
with open('example.txt') as f:
    content = f.read()

# If default encoding is 'utf-8' but file example.txt is encoded with 'latin-1', this will throw "UnicodeDecodeError: ’utf-8’ codec can’t decode bytes in position 118-123"

# So let's specify the correct encoding to fix it:

with open('example.txt', encoding='latin-1') as f:
    content = f.read()
````

### Doctests

- Exceptions are great for dealing with expected things that might go wrong.
- Before it comes to that, we should make sure that everything else goes right!
    * Test your code!
    * Write docstrings explaining what a function is supposed to do (which hopefully is exactly what it does) and how to use it!

````python
def gcd(m, n):
    '''
    Computes the greatest common denominator of m and n.
    '''
    return m % n
# This implementation is clearly wrong.
# If we use this gcd() function somewhere else, expecting it to work correctly, we have a problem.
````

Doctests are a convenient way of combining these two things in a formal way.

````python
def gcd(m, n):
    '''
    Computes the greatest common denominator of m and n.

    >>> gcd(8, 12)
    4
    >>> gcd(54, 24)
    6
    '''
    return m % n

if __name__ == '__main__':
    import doctest
    doctest.testmod(verbose=True)
````

It's important to write multiple tests, in case one of them gets the correct answer by accident:

```
File "<input>", line 5, in __main__.gcd
Failed example:
    gcd(8, 12)
Expected:
    4
Got:
    8
Trying:
    gcd(54, 24)
Expecting:
    6
ok!
```

Expected output behavior must be exactly as if you were running the interactive console (like in IDLE).

- Careful with strings and complex types!

````python
def hello():
    '''
    >>> hello()
    hello world
    >>> hello()
    'hello world'
    >>> print(hello())
    hello world
    '''
    return "hello world"

if __name__ == '__main__':
    import doctest
    doctest.testmod(verbose=True)
````
