
### The modulo operation

This operation computes the _remainder_ of a division. For example, consider dividing 7 by 3: $\frac{7}{3}$.
The result of the so-called __integer division__ is 2 in this case (the actual result of the division is 2.333), and the result of the modulo operation is 1.
The 3 fits into 7 only two times, and then 1 is left.
In other words: Imagine there are 7 apples and 3 students who want to share the apples. Each of the students wants to have as many apples as possible, but they have a strong feeling of fairness and want
to have the same number of apples each. They want to give the remaining apples to their Python tutor.
How many apples will the tutor receive? (That would be 7 modulo 3.)

The modulo operation actually doesn't care about the result of the division, but that is how it is derived. 
(The number of apples that the tutor receives depends only on the initial number of apples and how many students there are, not on how many apples each student receives.)
