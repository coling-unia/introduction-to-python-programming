By convention, class names should start with a capital letter. In order to create our account objects, we define an `Account` class. ## Methods

### Methods are functions that belong to a class

<div class="alert alert-block alert-info">

Methods are functions that belong to a class.<br/> __Instance methods__ are designed to operate on objects that have been created from this class.
The first parameter of an instance method is called `self` (by convention). This local variable of the method points to the object on which the method was called.
</div>

Two account objects have some commonalities, for instance we can withdraw money from either account or deposit money into either account. In OOP, such functionalities associated with objects are called _behaviors_ and programmed as _methods_. Methods are essentially functions that belong to a class, i.e., they reside in the namespace of a class. We do not have to write the `withdraw(amount)` and `deposit(amount)` methods separately for each object of the same class. Instead, we write the methods that are shared by all objects of a class once in the class definition, and the methods will be available for all objects that are created by calling this class. This is one reason why classes are so terribly useful.

````python
class Account:
    # Instance method
    def deposit(self, amount):
        self.balance += amount
    def withdraw(self, amount):
        self.balance -= amount

marcsAcc = Account()
marcsAcc.balance = 200
marcsAcc.deposit(500)

# Accessing attributes
print("Marc has", marcsAcc.balance, "Euros.")
````

The methods `deposit(self, amount)` and `withdraw(self, amount)` are _instance methods_ of the `Account` class. This means that they are designed to operate on objects that have been created from this class. Technically, instance methods are required to have a ﬁrst parameter which is called self (it could be called something else, too, but self is a very strong convention in Python). The following image shows that we have an object, `marcsAcc`, which is linked to the class `Account`. The `balance` attribute is only available at the particular object, but the methods that have been deﬁned in the class are available in the class.

<img src="/introduction-to-python-programming/images/09/oop10.png" alt="box labeled marcsAcc with an arrow pointing to a box labeled Account">


We can call the class’s methods "on an object" using the dot notation: `marcsAcc.deposit(500)`. In the code written for the class, however, the method looks like this: `deposit(self, amount)`. It has two parameters, but we called it only with one (amount). So why is that?

The following happens here: Python knows that `marcsAcc` is an object of the `Account` class (because we created the object by calling the class). Python executes the `deposit(self, amount)` method of the `Account` class and __assigns the object on which the method was called to the `self` parameter__. This means that the following two lines of code are equivalent:

```
marcsAccount.deposit(500)
Account.deposit(marcsAccount, 500)
```
When operating on instance objects, you should always use the ﬁrst way, because
it leaves the job to look up the class to Python and your code will be more ﬂexible.
This will become clear later when we talk about inheritance.

The following image illustrates the concept in details: When calling a method "on" an object using the dot notation (here, `deposit` is called on `marcsAcc`), the `deposit` method of the class from which `marcsAcc` was created is called. Inside the method, the parameter `self` points to the same object as `marcsAcc` points to. That's all the magic!

<img src="/introduction-to-python-programming/images/09/method_call-1.png" alt="image illustrating method call concept" width="600px"/>

Even easier to remember: Python starts to search for a method called on an object at the object itself. Technically, we can define methods for individual objects, but practically, methods are always coded in classes. Then, Python searches for the method in the class from which the object was created.

### Constructors are useful for initialization

The code for the class as written above is not very robust. It does not ensure that
the account object actually has a `balance` attribute before trying to print it out. If
we just add the following two lines of code to the main program, Python gives us an
error, complaining about the fact that we try to print out a `balance` attribute of the
`stefansAcc` object, although we never created a balance attribute for this object.

<div class="alert alert-block alert-info">
Remember to always make sure that any attribute that your methods try to access
actually exists. In this case, using a constructor method is helpful. In Python, a construc-
tor method is in fact an initialization method.
</div>

````python
class Account:
  # CONSTRUCTOR
  def __init__(self, iban, person):
    self.balance = 0
    self.iban = iban
    self.holder = person
  # METHODS
  def deposit(self, amount):
    self.balance += amount
  def print_info(self):
    print("Account holder:", self.holder)
    print("Balance:", self.balance)

# Main part of the program - execution starts here!
annesAcc = Account(1, "Anne")
annesAcc.deposit(42)
annesAcc.print_info()
````

In Python, constructor methods must have the name `__init__` (note the two underscores at either side), and it must have a self parameter. When creating an object, such as in line 15, the following happens: Python creates a new object, `annesAcc` and then calls the constructor method `__init__(self, iban, person)`. The values of the parameters `iban` and `person` are given when creating the object by calling the class: `Account(1, "Anne")`. Again, the new `annesAcc` object is automatically assigned to the `self` parameter, and inside the constructor method, we can now assign the necessary attributes to the object.

The Python statement `annesAcc = Account(1, "Anne")` triggers the following steps:

1. A new object is created from the `Account` class and assigned to the variable `annesAcc`.
2. The constructor method of the `Account` class is called. In this step, the newly created
object is assigned to the self parameter of the constructor method. So technically, Python executes `Account.__init__(annesAcc, 1, "Anne")`.
3. Inside the constructor method, the new object is initialized (the attributes are set to
the given or default values).

In Python, constructor methods do not _create_ an object. You can imagine that the
creation of an object happens "under the hood" and the constructor method then just
initializes the attributes of the object. (But this is rather a technical detail - just re-
member that the `__init__(self)` method is executed right after an object has been created.)


<div class="alert alert-block alert-info"> It is good programming practice to never modify (manipulate) instance attributes from "outside" the class code (a.k.a. client code -- the code that <i>uses</i> the class). </div>

<div class="alert alert-block alert-warning"> In Python, an attribute starting with an underscore signals that it is intended to be a <i>private</i> variable, i.e., one that should <i>REALLY</i> not be modified from outside as it would likely lead to the code breaking.<br/> &#9888; This is just a convention, not a security mechanism! </div>

````python
# Class definition code
class Account:
  def __init__(self, iban, person):
    self._balance = 0  # we are initializing _balance as a private attribute, because we do not want users to access it directly
    self.iban = iban
    self.holder = person
  def deposit(self, amount):
    self._balance += amount
````

## Class Design

Generally, when designing a class, you have to ask yourself the following questions:

<div class="alert alert-block alert-info"> 

1. How can I describe the state of my object? This will result in some _attributes_.
2. What do I know about the object before/when creating it? This goes into the
_initialization_ method.
3. What operations that change the object’s attributes will be performed on the
object? These operations need to be implemented as _instance methods_.
</div>

When designing the `Account` class, the answers to the above questions are as follows:

1. The state of an account is described by its account number, account holder
and the current balance.
2. When creating an account, I know the new account number and the holder’s
name. The default balance of an account is 0. We pass the account number
and the holder’s name to the constructor method, and set the default balance
to 0.
3. We will change the balance attribute of the account when withdrawing or
depositing money. Also, a method that prints out all information we have
about the state of the account is useful.

### Manipulate attributes only via instance methods

It was said earlier that assigning a value to an object’s attribute like this is bad style:
`stefansAcc.balance = 1000`.

In OOP, an important principle is that of _data encapsulation_, which means that
the attributes of an object should be "hidden" from manipulations from "outside" (i.e.
from the code that uses the object). The attributes of an object should only be modiﬁed
using code that was written within the class deﬁnition. This ensures that the state of
the object is always valid. Imagine, Stefan’s account balance is &euro;1000, and he wants to
withdraw &euro;1500. However, Stefan’s account has the restriction that it may not have a
negative balance. Assume the (human) teller at the bank forgets about this restriction,
sets the balance manually to -&euro;500 and hands the cash to Stefan. You can imagine that
this would make the branch manager of the bank quite unhappy when he realizes that
Stefan’s account is in a state that is not allowed. The `withdraw(self, amount)`
method modiﬁes the balance attribute, but we can control this manipulation in the
method. For instance, we could prohibit withdrawals that would result in a negative
balance.

````python
class Account(object):
    ''' a class representing an account '''
    def __init__(self, iban, person):
        self._balance = 0  # we are initializing _balance as a private attribute, because we need to treat it with care in withdraw() rather than having users access it directly
        self.iban = iban
        self.holder = person
    # METHODS
    def deposit(self, amount):
        self._balance += amount
    def withdraw(self, amount):
        if amount > self._balance:
            amount = self._balance
        self._balance -= amount
        return amount

# Main part of the program
annesAcc = Account(1, "Anne")
annesAcc.deposit(200)
print("Trying to withdraw 40:")
cash1 = annesAcc.withdraw(40)
print("Yeah, I got:", cash1)
print("Trying to withdraw 190:")
cash2 = annesAcc.withdraw(190)
print("Oh no, I only got:", cash2)

stefansAcc = Account(2, "Stefan")
stefansAcc.deposit(1000)
cash = stefansAcc.withdraw(1500)
print("Oh no, I only got:", cash)
````

### Changing attributes using setter methods

Sometimes, an attribute has to be changed completely. Assume, for tax reasons, it
is preferable for Stefan to change the account holder to its wife. Because we said
that assigning a value to an attribute from the outside like `stefansAcc.holder =
"Andrea"` is bad style, we provide a setter method for this case.

````python
class Account:
  def __init__(self, iban, person):
    self._balance = 0
    self.iban = iban
    self.holder = person
  def set_holder(self, person):
    self.holder = person

stefansAcc = Account(2, "Stefan") # Stefan opens an account
stefansAcc.set_holder("Andrea")   # The account is transfered to his wife's name
print("Account holder", stefansAcc.holder) # prints "Andrea"
````

<div class="alert alert-block alert-info"> 

The class code should provide a __setter__ method for each attribute whose value can be set from outside the client code. This lets you control what happens to the state of the object's data.
</div>

In the above code, the setter method simply sets the object’s attribute to the given value. This may seem superﬂuous at the moment, and there are actually other (safer) ways to do this in Python (keyword: properties). However, for now, we go along with the Python programming principle of trusting the code that uses your classes and stick to the following rules in order not to violate the data encapsulation paradigm:

<div class="alert alert-block alert-info"> 

1. Assign values to attributes only via instance methods (setters) or the constructor.
2. Modify the values of attributes only via instance methods.
3. Accessing (reading) the value of attributes like `print(stefansAcc.balance)` is okay.
</div>

So what good is that? For instance, inside the setter methods, we can validate the
new value, or raise an exception if the new value is not valid. The new value for the
account holder attribute has to be a name - or at least a non-empty string that contains
only letters and spaces. We could hence write the setter method like this:

````python
import re

class Account:
  def __init__(self, iban, person):
    self._balance = 0
    self._iban = iban
    self._holder = person

  # Example for setter method with validation
  def set_holder(self, person):
    if (not type(person) == str):
      raise TypeError
    if not re.match(r"\w+( \w+)*", person.strip()):
      raise ValueError
    self._holder = person

  # Example for setter method with modification
  def set_iban(self, iban):
      # Removes spaces
      self._iban = iban.replace(" ", "")
````

## Python Properties

<div class="alert alert-block alert-info"> It is good programming practice to never modify (manipulate) instance attributes from "outside" the class code (a.k.a. client code -- the code that <i>uses</i> the class). <b>Properties</b> are an elegant way to handle this in Python.</div>

Python properties are an elegant way to handle setter and getter methods in Python. The whole trick is that some functions (setters/getters) are called "under the hood" when the client code tries to set an attribute or when the client code tries to access an attribute of an object.

<div class="alert alert-block alert-warning">
You can signal with the <code>@property</code> <i>decorator</i> that some name within the classes' namespace is actually a getter method for the attribute <b>with the same name</b>.
</div>

&#x2757; In the example below, the expression `account.iban` evaluates to the return value of the function `iban(self)`, which is called under the hood.

<div class="alert alert-block alert-warning">
You can signal with the <code>@x.setter</code> <i>decorator</i> that some name within the classes' namespace is actually a setter method for the attribute <b>with the name x</b>.
</div>

&#x2757; This setter method is called whenever we try to set the value of a variable, i.e., also in the `__init__` method!

````python
class Account:
  def __init__(self, iban):
    self.balance = 0
    self.iban = iban # Here assign to "external" name:
                     # This triggers line 11.
  @property
  def iban(self):
    return self._iban

  @iban.setter
  def iban(self, new_iban):
    # Assign to "private" name here:
    self._iban = new_iban.replace(" ", "")

account = Account("DE11 0000 1111 0000 11")
print(account.iban) # This triggers the getter method for iban
````

To understand the example above, think about the order in which steps are executed:

1. The `__init__(self, iban)` method is called when the `account` object is created.
2. Inside the `__init__` method, `iban` is just a local variable, i.e., it resides in the local namespace of the method. We next create an attribute `self.iban` that resides in the namespace of the object.
3. As we have created two methods, both called `iban` but with two different _decorators_, Python knows that `iban` is indeed a property (attribute) of the class and that whenever someone tries to set the value of the `iban` attribute, it should call the method _decorated_ with `@iban.setter`. It does so also when we assign the value of the local variable `iban` in the `__init__` method to the object's variable `self.iban`. The setter method has two parameters: `self` (referring to the object) and a second parameter, which happens to be called `new_iban` here. The value on the right-hand side of  `self.iban = iban` gets assigned to this second parameter.
4. In the setter method for `iban`, we apply a special trick: we create one of the "private" attributes whose name starts with an underscore, `self._iban`. The value is that of `new_iban` (recall: what the user tried to assign to the property `iban`), but modified (if necessary: here, we removed the whitespaces from the IBAN).
5. Finally, when we try to access the value of the attribute/property `iban` as in `print(account.iban)`, Python knows that it should call the getter function for this property and returns its value.

This might seem complicated at first, but all you need to do is remember the steps taken. Go over this "recipe" a few times (on different days) if you feel unsure about it - I am sure you will soon get the hang of it!

### String representations of objects

Often, it is useful to have a meaningful string representation of an object. If we tell
Python to print an object, for instance `print(annesAcc)`, Python gives the cryptic
answer `<__main__.Account object at 0xb74faf4c>`. Instead, we would rather have a
string representation that really tells us what is going on with the attributes of the object, i.e., some information on its state. In Python, we can simply add an instance method that has the name `__str__(self)` (note the two underscores, no parameters besides `self`!) to the class. This method must return some string that describes the object.

```
class Account:
    ...
    def __str__(self):
        res = "*** Account Info ***\n"
        res += "Account ID:" + str(self.number) + "\n"
        res += "Holder:" + self.holder + "\n"
        res += "Balance: " + str(self.balance) + "\n"
        return res
```

In some cases, Python automatically realizes that we want a string representation
of an object, for instance when calling `print(annesAcc)` or `str(annesAcc)`, it
automatically calls the `__str__(self)` method of this object.
When calling `print(annesAcc)`, it prints out whatever the `__str__(self)` method
of the `annesAcc` object returns; `str(annesAcc)` returns the string that is returned
by the method call `annesAcc.__str__()`.

_Hooks_ are special functions of a class that are called automatically by Python in
some circumstances. The `__str__(self)` method is only one example. There
is another hook method called `__repr__(self)` which is also supposed to return a
string that describes the object. `__str__(self)` is supposed to return a user-friendly
description of the object, while `__repr__(self)` should return a description that is
useful for developers. Note that `print(anObject)` returns the string returned by
the `__str__(self)` method, while printing an object by just typing it in the interactive
mode returns the string returned by `__repr__(self)`. Another hook function that
we have seen already is the constructor method (`__init__(self, args)`), which is
called when creating a new object by calling a class. There are also hook methods
that are triggered when operators are applied, for example the operator + triggers the
add method of an object. Note that the names of hook methods usually start and
end with two underscores. We will learn more about hooks later.


<div class="alert alert-block alert-info"> 

There are two useful "magic" functions (_hooks_) that are executed automatically by Python in particular circumstances to get information on objects:

* `__str(self)__` is called when a string representation of the object is requested, i.e., by when the client code says `str(some_object)` or `print(some_object)`. This function should provide a simple description for the user, or when type conversion to a string is required (compare: what does `str(5)` return?).
* `__repr(self)__` is similar, but should conventionally return a detailed description that could be useful to the programmer, e.g., for logs.
</div>

````python
class Account:
  
  def __init__(self, iban, holder):
    self.balance = 0
    self.iban = iban
    self.holder = holder
  
  def deposit(self, amount):
    self.balance += amount

  def __str__(self):
    res = "*** Account Info ***\n"
    res += "Account ID:" + str(self.iban) + "\n"
    res += "Holder:" + self.holder + "\n"
    res += "Balance: " + str(self.balance) + "\n"
    return res
  
  def __repr__(self):
    class_name = type(self).__name__
    return f"{class_name}(holder={self.holder}, balance={self.balance})"
    
marcsAcc = Account("DE11 0000 1111 0000 11", "Marc")
marcsAcc.deposit(200)
print("__str__")
print(marcsAcc)

print("\n__repr__")
print(repr(marcsAcc))
````

## Composition and Aggregation

Recall that each value in Python has a type, e.g. `1.5` has the type ﬂoat or `"python"` has the type str (=string). If we create an object, it also has a type, which is the class from which it was created.

The attributes of an object can be of any type. This means that the attributes of an object can be objects themselves. In fact, although we haven’t told you so far, anything is an object in Python, also strings, numbers, lists or dictionaries. The terms _composition_ and _aggregation_ refer to the fact that we can compose more complex objects out of multiple objects.

* _Composition_ implies that the owned objects only exist "within" the more complex object. For instance, if a university is closed, so are all its departments.
* _Aggregation_ implies that an object is part of a more complex object, but if the larger object is destroyed, the smaller one may still exist. For instance, we can deﬁne a `Person` class that deals with information about persons, and we save the information about an account holder in a `Person` object as shown in the following example. If the account is closed, the person still exists (and so may its object - a bank might save information about a former customer even after the customer closed his or her account).

This ability to create objects out of objects is called _compositionality_.
We access objects or values residing inside other objects (like namespaces) with dot notation: `marcsAcc.holder.name`.
Don’t make these calls too long (for readability and security)!

<img src="/introduction-to-python-programming/images/09/composition-1.png" alt="image illustrating with boxes that some objects can be the attribute value of another object" width="600px" />

````python
class Person:
    def __init__(self, f, l, a):
        self.firstname = f
        self.lastname = l
        self.age = a
  
class Account:
    def __init__(self, person, num):
        self.holder = person
        self.num = num
        self.balance = 0
    def deposit(self, amount):
        self.balance += amount

# Compositionality  
marc = Person("Marc", "Schulder", 29)
marcsAcc = Account(marc, 1)
marcsOtherAcc = Account(marc, 2)

print(marcsOtherAcc.holder.age)
marcsAcc.holder.age += 1
print(marcsOtherAcc.holder.age)
````

Compositionality is extremely useful and at the core of OOP. However, we need to be aware of where attributes point to: in our example, `marcsAcc.holder.age += 1` also changes `marcsOtherAcc.holder.age`. This may be desired here, but possibly leads to bugs in another setting.

<img src="/introduction-to-python-programming/images/09/composition2-1.png" alt="image illustrating side effects that may occur in complex objects as the same value may be incremented through various references (directly or via attribute)" width="600px"/>



### Composition vs. Aggregation

__Composition__ = contained objects only exist within a complex object.

_Example_: Object `balance` (of type `int`) does not exist without an `Account` object. If the `Account` object is deleted, we do not have any reference to the balance any more.

__Aggregation__ = no implied ownership; contained objects can exist independently; can also be used in more than one object.

_Example_: An `Account` object has a reference (attribute) to a `Customer` object. If the `Account` object is deleted, the application may still have a reference to the `Customer` object, e.g. if the customer has more than one account.

Composition can not really be enforced in Python, as object
attributes can not be hidden from the user.

In UML class diagrams, we can show that one class has an attribute whose type is another class using a line that connects the two classes. The diamond symbol shows that the `Account` class uses the `Person` class. In the case of composition, the diamond symbol is filled (black).

<img src="/introduction-to-python-programming/images/09/python-uml-diagram.png" alt="UML diagram with two boxes (which represent classes) indicating aggregation using a diamond on the link between the two boxes" width="600px" />

Please note that UML class diagrams show the class design, i.e. how objects created from this class will look like. They show the attributes and methods that will be available for objects created from this class. UML diagrams do not show what is going on in Python internally (for this we used the informal diagrams with the blue/grey boxes). For example, instance attributes will not be available in the class object, but we list them in the class diagram. Class attributes and static methods are underlined in UML class diagrams.
