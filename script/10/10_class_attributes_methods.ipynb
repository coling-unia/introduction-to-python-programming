We will now introduce some more advanced features of object-oriented programming in Python.

In Python, classes are objects, too. They are created when defining a class using the `class` statement. After defining the `Account` class, exactly one _class object_ for the `Account` class becomes available. Each time we call this class (e.g. `annesAcc = Account()`), we create a new \textit{instance object} of this class. The object `annesAcc` is automatically linked to the `Account` class object; we say `annesAcc` 'is-a' `Account`, because it's an instance of the `Account` class. 

When we try to access a component (an attribute or a method) of an object using the dot notation in the way `objectName.componentName`, Python first looks in the object itself. Each object can be seen as its own namespace. If Python finds the `componentName` there, it is happy. If it doesn't, it looks in the class object of the class from which the object was created. If it doesn't find the `componentName` there, an error occurs.

Let's look at an example. Assume that the `Account` class has three methods, `withdraw(self, amount)`, `deposit(self, amount)` and `print_info(self)`. We created the object `annesAcc` by calling the `Account` class, and then we assigned an attribute to the `annesAcc` object by writing `annesAcc.balance = 200`. The `annesAcc` object is linked to the class it was created from, as shown in the image below.

## Class Attributes and Shadowing

We already know that classes are some kind of objects, too. A class object can also have attributes, which we call _class attributes_. In the following example, the `Account` class has a class attribute called `num_of_accounts`, which records how many accounts were created.

<div class="alert alert-block alert-info"> Recall: classes are objects, too.
Hence, they can have attributes, too: <i>class attributes</i>. </div>

````python
# Class Attributes
Account.num_of_accounts = 0
print(Account.num_of_accounts)
````


So what exactly happens here? We create the class attribute by assigning a variable called `num_of_accounts` in the namespace of the `Account` class. Note that the class variable has the same indent as the methods we defined within the class. At the time when the class object is created, the initial value for the class attribute is set to 0.

<div class="alert alert-block alert-info"> A class attribute's value can be set,
for example, in the class definition code in the namespace of the class (outside any method). This is executed when the class definition code is read in. </div>


````python
# Using and Accessing Class Attributes
class Account:
    num_of_accounts = 0
    def __init__(self, iban, holder):
        self.iban = iban
        self.holder = holder
        self.balance = 0
        Account.num_of_accounts += 1

print(Account.num_of_accounts) # 0 because the class definition code has just been executed
marcsAcc = Account("DE56", "Marc")
print(Account.num_of_accounts)
annesAcc = Account("DE57", "Anne")
print(Account.num_of_accounts)
print(annesAcc.num_of_accounts)
````


Class attributes are usually initialized in the class definition code before the `__init__` method.
In the constructor method, we modify the `num_of_accounts` class attribute.
Recall that the `__init__` method is called each time when an instance object is created from the class. Each time this happens, we increment the `num\_of\_accounts` class attribute. We access the class attribute by using the class name together with the dot notation: `Account.num_of_accounts`.

The image on the previous page shows that the `num_of_account` attribute resides within the `Account` class object. Up to now, we have seen the concept: Class objects can have attributes, too, and we assign them in the namespace of the class (either in the class with the correct indent or using `ClassName.attributeName`). However, we can also access the class attributes via instance objects that were created from the class, as the interactive session shows:

 ```
>>> Account.num_of_accounts
2
>>> annesAcc.num_of_accounts
2
>>> stefansAcc.num_of_accounts
2
```

So far, so good. In the following case, however, it gets tricky and we have to program carefully in order to avoid bugs.
This is because we can _read_ class attributes via object instances, but we _cannot change_ them this way!

Whenever we assign something to a variable in namespace, it becomes local to this namespace. This is the same principle as we have learned about when introducing functions and namespaces! If you are unsure about this, now is a good time to review Chapter 7 on Functions and Namespaces!

In the example below, when expanding the `+= 5` shorthand, it becomes visible that we assign something to a new local variable of `annesAcc` that now "shadows" `Account.num_of_accounts` which was previously found when looking for `num_of_accounts` in `annesAcc`.

````python
# Careful: Accessing Class Attributes --> Shadowing!
print(annesAcc.num_of_accounts)
annesAcc.num_of_accounts += 5 # corresponds to annesAcc.num_of_accounts = Accounts.num_of_accounts + 5
print(Account.num_of_accounts)
print(annesAcc.num_of_accounts)
````

<img src="/introduction-to-python-programming/images/09/class_attributes1.png" alt="the same image as above, but now annesAcc has an attribute called num_of_accounts as well" width="400px">

This is called _shadowing_: Creating instance attributes with the same name as class attributes means that the class attribute can not be accessed through the instance object anymore, as the instance attribute blocks it.

````python
class Account:
    # Class Attributes
    num_of_accounts = 0
    # Initialization
    def __init__(self, iban, holder):
        self.iban = iban
        self.holder = holder
        self.balance = 0
        Account.num_of_accounts += 1

marcsAcc = Account("DE56", "Marc")
Account.num_of_accounts # == 1; Class attribute
marcsAcc.num_of_accounts # == 1; Class attribute
marcsAcc.num_of_accounts = 7
Account.num_of_accounts # == 1; Class attribute
marcsAcc.num_of_accounts # == 7; Instance attribute
````

An easy way to avoid this pitfall is to give distinct names to your class attributes and never use the same names for instance attributes.


<div class="alert alert-block alert-warning"> It is not recommended to use the
same attribute name for a class attribute and for an instance attribute.
The class attribute can be accessed via the instances, but if an instance attribute is created with the same name (as in the following example), this instance attribute can <i>shadow</i> the class attribute. </div>


<div class="alert alert-block alert-danger"> Never use the same (shadowing) name for a class attribute and an instance attribute. Always access class attributes via the class name, never access it from the instance object.</div>

## Class Methods

So far, we have only seen _instance methods_, which we define inside a class, but which we can only call 'on an object', i.e. when a particular instance is involved. We will now learn about two additional types of methods a class can have.
The first is called _class method_ and it belongs purely to the class, i.e., it does not have access to an instance object via a `self` parameter. However, class methods have direct access to the _class object_ itself via a _cls_ parameter. Class methods can hence modify the state of their class.

<div class="alert alert-block alert-info"> A <strong>class method</strong> (which "belongs" to the class object, not the instances of the class) is declared using the <code>@classmethod</code> decorator. It has the attribute <code>cls</code> that points to the class object. </div>

````python
# Class Methods
class Account:
    num_of_accounts = 0
    # Initialization
    def __init__(self, iban, holder):
        self.iban = iban
        self.holder = holder
        self.balance = 0
        Account.num_of_accounts += 1
    # Instance methods
    def deposit(self, amount):
        self.balance += amount
    
    @classmethod # <-- decorator
    def reset_accounts(cls):
        cls.num_of_accounts = 0
````

__Use case for class methods__

A __factory__ is a [design pattern](https://www.pearson.de/design-patterns-elements-of-reusable-object-oriented-software-9780201633610) that helps to create objects from a class, or variants of that class. Here, we simply instantiate objects of the same class with different pre-defined interest rates by calling one of two class methods.

__Hint__: It is not required that you use class methods in your own code - there are many ways to implement a solution to a problem. However, you should know what they are and be able to read code using them.

````python
# Example Use Case for Class Methods: Factories
class Account:
    def __init__(self, interest_rate):
        self.interest_rate = interest_rate

    @classmethod
    def savings_account(cls):
        return cls(0.03)

    @classmethod
    def checking_account(cls):
        return cls(0.00)
    
    def __str__(self):
        s = str(type(self)) + " interest_rate=" + str(self.interest_rate)
        return s

annesAccount = Account.savings_account()
print(annesAccount)
````

## Static Methods

A class can also have methods that do not involve particular instances, but which are called on the class object. Such methods are called _static methods_.

When defining a static method, we need to write `@staticmethod` in the line before the `def` statement for the respective method. `@staticmethod` is a `decorator`. For now, you can imagine that it is a special message you pass to Python, here telling it that the method you are defining belongs to the class (not to particular objects created from the class). The method also does not have the `self` parameter like the instance methods we have seen before.


<div class="alert alert-block alert-info"> A <strong>static method</strong> is declared using the decorator <code>@staticmethod</code>. It can modify neither the object state nor the class state, it may take any number of regular parameters. Instead of using a static method, you can also just use a regular function. Static methods provide primarily a way of grouping your code in a namespace. Often, the namespace of your module (assumed it's informative) will and a regular function can do the same job!</div>

````python
# Static methods
class Account:
    num_of_accounts = 0

    def __init__(self):
        Account.num_of_accounts += 1

    @staticmethod
    def accounts_info(username):
        print("Info for", username, ":")
        print(Account.num_of_accounts, "accounts have been created.")

a = Account()
b = Account()
c = Account()
Account.accounts_info("Anne")
````

### Summary

* Both class and static methods can be called "on" an object (not recommended).
* Instance methods have access to the class via the `self.__class__` attribute.

## Python Modules

Usually, you will write your classes into various ﬁles (_modules_), and you want to use them in another ﬁle which contains your main application. An easy way to be able to use your class is to write the class deﬁnition into a ﬁle (module) which we call `accounts.py`, and then have the main part of your application in another ﬁle, here called `bank.py`. The `bank.py` ﬁle must then import your classes. Note that the syntax for importing your class is `from modulename import classname`, where `modulename` has to match the ﬁle name in which your class is deﬁned (minus the `.py`) and `classname` is the name of your class.

Modules are just files of Python code, can define functions, variables, classes and/or runnable code. Module name = filename minus `.py`.
Modules group related code, which makes code easier to understand.
When defining a function/class, Python creates function/class objects.
Modules are also objects, and they contain references to the function/class objects that the module defines. We can import these functions/classes in other modules (= Python files).

```
# import all functions / classes from <modulename>
import modulename

# import specific function
from modulename import somefunction

# import specific class
from modulename import someclass
```

* When importing a module, the functions / classes are available.
* Running a module: `python3 someModule.py`
* `someModule` may import code from other files.
* `__name__ == "__main__"` is `True` when the module is run, but `False` when it is imported. We can use this in each module for tests, but at runtime, only the main module defines what happens!

```
# imports

# function / class definitions

# main
if __name__ == "__main__":
    # this code is executed when running THIS module
```
