## Functions

As an example let us have a look at the factorial again.

n! = n · (n−1) · (n−2) · ... · 3 · 2 · 1

````python
x = 34
r = 1
for i in range(x):
    r *= (i+1)

print("The factorial of", x, "is", r)
````

But what if we want to reuse this logic? For that we have functions.

Functions are "subprograms" that can (and should) be used to divide a larger problem into several smaller problems. In our example it would like the following.

````python
def factorial(x):
    """Computes the factorial of x"""
    r = 1
    for i in range(x):
        r *= (i+1)
    return r
````

## Anatomy of a Function

:::{admonition} Syntax
:class: note

```py
def name(var_1, ..., var_n):
    """A short documentation (optional)"""
    <code>
    return <something>
```

- `name`: the name of the function (a variable)
- `var_1, ..., var_n`: the parameters of the function
- `return <something>`: usually at the end of the function defintion, optional

:::

## Function Defintion

The function defintion is similar to the assignement of function logic  to a variable. The function name can be treated like a variable.

````python
def factorial(x):
    """Computes the factorial of x"""
    r = 1
    for i in range(x):
        r *= (i+1)
    return r

print(factorial) # <function factorial at 0x1e57b0>
help(factorial) # Computes the factorial of x
````

## Function Application

To execute the code within the function definition the function is applied using the following syntax. 

````python
def factorial(x):
    """Computes the factorial of x"""
    r = 1
    for i in range(x):
        r *= (i+1)
    return r

print(factorial(14))
print(factorial(0))
````

- When the function is called, the parameters are instantiated with the values from the function call (more specific: _call-by-object-reference_, more on this later!).
- The function call evaluates to the value returned by the function.

- The `return` statement stops the execution of the function and returns a value (or a reference to a value).
